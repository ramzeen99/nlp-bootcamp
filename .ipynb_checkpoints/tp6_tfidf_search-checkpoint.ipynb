````python
# Mini-corpus de 15 résumés d'articles techniques
documents = [
    "Machine learning is a subset of artificial intelligence that enables systems to learn from data without explicit programming.",
    "Deep learning uses neural networks with many layers to learn complex patterns from large amounts of data.",
    "Natural language processing is a field of AI focused on interaction between computers and human language.",
    "The transformer architecture uses self-attention mechanisms to process sequential data in parallel.",
    "Word embeddings are dense vector representations of words that capture semantic meaning.",
    "Convolutional neural networks are specialized for processing grid-like data such as images.",
    "Recurrent neural networks process sequential data by maintaining a hidden state over time.",
    "Attention mechanisms allow models to focus on relevant parts of the input when making predictions.",
    "Python is a high-level programming language known for its readability and extensive libraries.",
    "Data science combines statistics, programming, and domain expertise to extract insights from data.",
    "Computer vision enables machines to understand and interpret visual information from images.",
    "Reinforcement learning trains agents to make decisions by rewarding desired behaviors.",
    "Transfer learning reuses pre-trained models on new tasks to reduce training time and data needs.",
    "Large language models like GPT use transformer architectures trained on massive text corpora.",
    "Generative adversarial networks consist of two neural networks competing against each other."
]

titles = [
    "Machine Learning",
    "Deep Learning",
    "NLP",
    "Transformer",
    "Word Embeddings",
    "CNN",
    "RNN",
    "Attention Mechanism",
    "Python",
    "Data Science",
    "Computer Vision",
    "Reinforcement Learning",
    "Transfer Learning",
    "Large Language Models",
    "GANs"
]

print(f"Total : {len(documents)} articles")
````

````python
# TF-IDF Vectorizer
vectorizer = TfidfVectorizer(
    max_features=5000,      # Garder 5000 mots max
    stop_words='english',    # Supprimer les stopwords anglais
    ngram_range=(1, 2)       # Unigrams + bigrams
)

# Transformer les documents en matrice TF-IDF
tfidf_matrix = vectorizer.fit_transform(documents)

print(f"Matrice TF-IDF : {tfidf_matrix.shape}")
# Ex : (12, 5000) = 12 documents, 5000 features
````

````python
def search(query, top_k=3):
    """
    Cherche les documents les plus pertinents pour une requête.
    
    Args:
        query  : string de recherche
        top_k  : nombre de résultats à retourner
    
    Returns:
        Liste de (titre, score) triée par pertinence
    """
    # 1. Vectorizer la requête
    query_vec = vectorizer.transform([query])
    
    # 2. Calculer la similarité cosinus entre la requête et tous les docs
    similarities = cosine_similarity(query_vec, tfidf_matrix)[0]
    
    # 3. Trier par score décroissant
    top_indices = np.argsort(similarities)[::-1][:top_k]
    
    # 4. Retourner les résultats
    results = []
    for i in top_indices:
        if similarities[i] > 0:  # Ignorer les scores nuls
            results.append((titles[i], similarities[i]))
    
    return results
````

````python
test_queries = [
    "how does attention work in transformers",
    "word vectors and embeddings",
    "deep learning for image recognition",
    "training neural networks with backpropagation",
    "python programming language"
]

for query in test_queries:
    print(f"\n🔍 Requête : '{query}'")
    print("-" * 50)
    results = search(query, top_k=3)
    for i, (title, score) in enumerate(results):
        print(f"  {i+1}. {title} (score: {score:.3f})")
````

````python
def explain_query(query):
    """Affiche les mots TF-IDF les plus importants d'une requête."""
    query_vec = vectorizer.transform([query])
    feature_names = vectorizer.get_feature_names_out()
    
    # Récupérer les indices des mots présents dans la query
    indices = query_vec.nonzero()[1]
    
    print(f"Requête : '{query}'")
    print("Mots importants :")
    for i in indices:
        print(f"  {feature_names[i]}: {query_vec[0, i]:.4f}")

explain_query("attention mechanism in transformers")
````

# 📋 RÉSUMÉ TP6 — Moteur de Recherche TF-IDF

---

## 🧠 LA MENTALITÉ

> **TF-IDF** pondère chaque mot par sa rareté dans le corpus. Les mots communs (`the`, `is`) sont automatiquement ignorés, les mots rares et pertinents (`transformer`, `embedding`) sont amplifiés. C'est la base de tout moteur de recherche classique.

---

## 📖 LES CONCEPTS CLÉS

### TF = Term Frequency
Fréquence d'un mot dans **un document spécifique**.

```
TF("deep", doc) = nombre d'occurrences de "deep" / nombre total de mots dans le doc
```

### IDF = Inverse Document Frequency
Rareté d'un mot dans **tout le corpus**.

```
IDF("deep") = log(N / nombre de docs contenant "deep")
```

- `the` apparaît dans tous les docs → IDF proche de 0 → ignoré
- `transformer` apparaît dans 1 doc → IDF élevé → amplifié

### TF-IDF = TF × IDF
Score final d'un mot dans un document.

### Similarité Cosinus
Mesure l'angle entre 2 vecteurs.

```
cos(A, B) = (A·B) / (||A|| × ||B||)
```

- 1 = même direction = très similaire
- 0 = orthogonal = pas similaire

---

## 🔍 LE FONCTIONNEMENT DU MOTEUR

```
1. Corpus de documents → TfidfVectorizer → Matrice TF-IDF
2. Requête utilisateur   → TfidfVectorizer → Vecteur TF-IDF
3. Cosinus(requête, chaque document) → Scores de similarité
4. Trier les scores → Top K résultats
```

### Vectorizer TF-IDF
```python
vectorizer = TfidfVectorizer(
    max_features=5000,      # Garde les 5000 mots/ngrams les plus fréquents
    stop_words='english',   # Supprime "the", "a", "is"...
    ngram_range=(1, 2)      # Capture "neural network" comme un seul token
)

# Transforme tous les documents en une matrice
# Lignes = documents, Colonnes = mots/ngrams, Valeurs = scores TF-IDF
tfidf_matrix = vectorizer.fit_transform(documents)
# Ex : (15, 241) = 15 docs, 241 features uniques
```

### Fonction search() — Cœur du TP
```python
def search(query, top_k=3):
    # 1. Vectoriser la requête (même vocabulaire que les docs)
    query_vec = vectorizer.transform([query])
    # query_vec est un vecteur de 241 nombres TF-IDF
    
    # 2. Calculer la similarité cosinus entre la requête et CHAQUE doc
    similarities = cosine_similarity(query_vec, tfidf_matrix)[0]
    # similarities = [0.0, 0.0, 0.35, 0.12, ...] — un score par document
    
    # 3. Trier les indices par score décroissant
    top_indices = np.argsort(similarities)[::-1][:top_k]
    # argsort = indices qui trient le tableau
    # [::-1] = ordre décroissant
    # [:top_k] = garder les k premiers
    
    # 4. Retourner les titres et scores > 0
    results = []
    for i in top_indices:
        if similarities[i] > 0:
            results.append((titles[i], similarities[i]))
    return results
```

## 2. LE PROBLÈME AVEC LES MOTS FRÉQUENTS

Imagine 3 articles :

| Article | Contenu |
|---------|---------|
| A | "the cat is on the mat" |
| B | "deep learning neural networks" |
| C | "python is a programming language" |

Tu cherches `"deep learning"`. L'article B devrait gagner. Mais si on compte juste les mots :

- `the` apparaît 2 fois dans A, 0 dans B
- `is` apparaît dans A et C

Résultat : A pourrait gagner juste parce qu'il a `the` et `is`, qui n'ont rien à voir avec ta recherche.

👉 **TF-IDF règle ce problème.**

---

## 3. TF-IDF : LES 2 COMPOSANTES

### TF = Term Frequency (Fréquence du mot dans ce document)

```
TF("deep", article B) = 1/4 = 0.25
```

Plus un mot apparaît dans un document, plus son TF est élevé.

### IDF = Inverse Document Frequency (Rareté du mot dans tout le corpus)

```
IDF("the") = log(3/2) = 0.18   ← "the" est dans 2 docs sur 3 → banal
IDF("deep") = log(3/1) = 0.48  ← "deep" est dans 1 doc sur 3 → rare !
```

Plus un mot est rare dans l'ensemble du corpus, plus son IDF est élevé.

### TF-IDF = TF × IDF

| Mot | TF (dans B) | IDF (corpus) | TF-IDF |
|-----|-------------|--------------|--------|
| `the` | 0 | 0.18 | 0 |
| `deep` | 0.25 | 0.48 | **0.12** |
| `learning` | 0.25 | 0.30 | **0.08** |

👉 Les mots rares et pertinents ont un score élevé. Les mots communs ont un score proche de 0.

---

## 4. SIMILARITÉ COSINUS

Une fois que chaque document et ta requête sont transformés en vecteurs TF-IDF, on calcule **l'angle** entre eux :

```
cos(requête, document) = (requête · document) / (||requête|| × ||document||)
```

- cos = 1 → même direction → très pertinent
- cos = 0 → orthogonal → pas pertinent du tout
